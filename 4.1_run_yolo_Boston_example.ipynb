## 4.1. Run YOLO

This notebook shows a tree object detection pipeline to add tree position bounding boxes to the dataset.
Later, these tree bounding boxes are used to compute the region directly below the tree

```python
from ultralytics import YOLO
import numpy as np
# Load an Open Images Dataset V7 pretrained YOLOv8n model
# Source: https://docs.ultralytics.com/datasets/detect/open-images-v7/#open-images-v7-pretrained-models
# Citation: Jocher, G., Qiu, J., & Chaurasia, A. (2023).
# Ultralytics YOLO (Version 8.0.0) [Computer software]. https://github.com/ultralytics/ultralytics
model = YOLO("yolov8n-oiv7.pt")
```

```python
def produce_rectangle(image_path):
    results = model.predict(source=image_path)
    names = [results[0].names[cls.item()] for cls in results[0].boxes.cls.int()]
    index_of_tree = names.index('Tree')
    xyxy = np.array(results[0].boxes.xyxy[index_of_tree].cpu())
    xyxy = [int(x) for x in xyxy]
    
    return xyxy
    
```

```python
import pandas as pd
import cv2, os
os.chdir('/home/klimenko/CoolingMachines/explore_flir')
df = pd.read_csv('dataset/AMSTERDAM_FLIR_DATASET/all_26.csv')

# Initialize empty lists to hold the computed values
xyxy_values = []

# Loop over the rows in the DataFrame and apply the function
total_rows = len(df)
for idx, row in df.iterrows():
    
    try:
        # Extract values using the image path and info

        xyxy = produce_rectangle(row['visual_path'])


        # Append the results to the respective lists
        xyxy_values.append(xyxy)


        # Print progress manually
        if (idx + 1) % 10 == 0 or (idx + 1) == total_rows:
            print(f"Processed {idx + 1}/{total_rows} rows____________________")
            
    except:
        print('F')
        xyxy_values.append(np.nan)


# Add the computed values as new columns in the DataFrame
df['xyxy'] = xyxy_values
```

```python
df.to_csv('dataset/AMSTERDAM_FLIR_DATASET/all_26.csv')
```
